## Purpose

Baby Step 0 establishes the exo-brain's durable memory before asking it to make any strategic judgment.

The vault contains two distinct synthetic universes:

1. **A precedent corpus of 1,000 synthetic judicial decisions**
2. **A working portfolio of five active corporate civil disputes**

The five active matters are not precedents. They are the disputes for which later baby steps will develop and revise strategy.

No recommendation is produced in Baby Step 0. The permitted state is deliberately narrow:

- observe;
- organize;
- browse;
- inspect relationships;
- test the structure.

No litigation strategy, filing, communication, settlement decision, or external action is authorized.


## Why the dual-universe architecture matters

The exo-brain needs to distinguish between:

- the legal environment;
- the active disputes being managed;
- later analytical conclusions;
- human decisions.

The **precedent corpus** represents the legal environment.  
The **active matters** represent the five current problems.  
Later recommendation records will represent what the institution believes should be done.

This separation prevents a recommendation from being confused with law and prevents a synthetic precedent from being confused with a client fact.

## Synthetic precedent corpus

The initial corpus contains 1,000 fictional judicial decisions.

The corpus is deliberately heterogeneous:

- seven jurisdictions;
- multiple court levels;
- ten recurring corporate civil causes of action;
- different procedural postures;
- binding and persuasive authority;
- positive, neutral, distinguished, criticized, and overruled treatment states.

The goal is architectural, not predictive. The synthetic corpus allows the exo-brain to be tested without exposing confidential legal material or implying real legal advice.


```python
RANDOM_SEED = 20260720
random.seed(RANDOM_SEED)

JURISDICTIONS = [
    "New York", "Delaware", "California", "Texas",
    "Florida", "Illinois", "Massachusetts"
]

COURTS = {
    "New York": [
        "New York Court of Appeals",
        "New York Appellate Division",
        "New York Commercial Division"
    ],
    "Delaware": [
        "Delaware Supreme Court",
        "Delaware Court of Chancery",
        "Delaware Superior Court"
    ],
    "California": [
        "California Supreme Court",
        "California Court of Appeal",
        "California Superior Court"
    ],
    "Texas": [
        "Texas Supreme Court",
        "Texas Court of Appeals",
        "Texas District Court"
    ],
    "Florida": [
        "Florida Supreme Court",
        "Florida District Court of Appeal",
        "Florida Circuit Court"
    ],
    "Illinois": [
        "Illinois Supreme Court",
        "Illinois Appellate Court",
        "Illinois Circuit Court"
    ],
    "Massachusetts": [
        "Massachusetts Supreme Judicial Court",
        "Massachusetts Appeals Court",
        "Massachusetts Superior Court"
    ]
}

CAUSES = {
    "Breach of Shareholders Agreement": [
        "injunctive relief", "transfer restriction", "buy-sell mechanism"
    ],
    "Post-M&A Purchase Price Dispute": [
        "working-capital adjustment", "earnout interpretation", "expert determination"
    ],
    "Technology Licensing Dispute": [
        "scope of license", "royalty calculation", "confidentiality"
    ],
    "Joint Venture Deadlock": [
        "deadlock remedy", "fiduciary duty", "dissolution"
    ],
    "Supply Agreement Dispute": [
        "force majeure", "consequential damages", "mitigation"
    ],
    "Commercial Misrepresentation": [
        "materiality", "reliance", "causation"
    ],
    "Breach of Confidentiality": [
        "protected information", "misuse", "irreparable harm"
    ],
    "Corporate Governance Dispute": [
        "board authority", "fiduciary duty", "minority protection"
    ],
    "Construction and Development Dispute": [
        "delay", "defect", "liquidated damages"
    ],
    "Commercial Lease Dispute": [
        "default", "termination", "damages"
    ]
}

PROCEDURAL_POSTURES = [
    "Motion to Dismiss",
    "Preliminary Injunction",
    "Summary Judgment",
    "Post-Trial Judgment",
    "Interlocutory Appeal",
    "Final Appeal",
    "Arbitration Enforcement",
    "Discovery Sanctions"
]

PRECEDENTIAL_STATUS = [
    "Binding",
    "Persuasive",
    "Limited Persuasive",
    "Unpublished",
    "Arbitral Persuasive"
]

TREATMENT = [
    "Positive",
    "Followed",
    "Distinguished",
    "Criticized",
    "Limited",
    "Overruled"
]

DISPOSITIONS = [
    "Affirmed",
    "Reversed",
    "Vacated",
    "Remanded",
    "Granted",
    "Denied",
    "Partially Granted"
]

def level_weight(court_name):
    if "Supreme" in court_name or "Court of Appeals" in court_name or "Judicial Court" in court_name:
        return 3
    if "Appellate" in court_name or "Appeals Court" in court_name or "Court of Appeal" in court_name:
        return 2
    return 1

precedents = []

for i in range(1, 1001):
    pid = f"PRE-{i:04d}"
    jurisdiction = random.choice(JURISDICTIONS)
    court = random.choice(COURTS[jurisdiction])
    cause = random.choice(list(CAUSES.keys()))
    issue = random.choice(CAUSES[cause])
    year = random.randint(2000, 2026)
    decision_date = datetime.date(
        year,
        random.randint(1, 12),
        random.randint(1, 28)
    )

    precedents.append({
        "precedent_id": pid,
        "caption": f"Atlas Corporate {i:04d} v. Meridian Enterprise {i:04d}",
        "court": court,
        "court_level": level_weight(court),
        "jurisdiction": jurisdiction,
        "decision_date": decision_date.isoformat(),
        "precedential_status": random.choice(PRECEDENTIAL_STATUS),
        "procedural_posture": random.choice(PROCEDURAL_POSTURES),
        "cause_of_action": cause,
        "legal_issue": issue,
        "material_facts": (
            f"A synthetic {cause.lower()} dispute involving "
            f"{issue.lower()} and contested corporate conduct."
        ),
        "holding": (
            f"The court issued a synthetic holding concerning {issue.lower()} "
            f"under {jurisdiction} law."
        ),
        "rule_of_law": (
            f"Illustrative rule: resolution of {issue.lower()} depends on "
            f"contract language, procedural posture, and evidentiary support."
        ),
        "disposition": random.choice(DISPOSITIONS),
        "authorities_cited": [],
        "authorities_distinguished": [],
        "authorities_criticized": [],
        "treatment_status": random.choice(TREATMENT),
        "synthetic": True
    })

assert len(precedents) == 1000
print("Synthetic precedents created:", len(precedents))
print(json.dumps(precedents[0], indent=2))
```

## Citation and treatment relationships

A precedent corpus is not merely a document collection. Judicial decisions interact through citation and treatment.

The notebook therefore creates a synthetic relationship graph in which cases may:

- cite;
- distinguish;
- criticize;
- limit;
- or overrule earlier authorities.

Only earlier-dated precedents may be cited by later decisions.


```python
precedents_sorted = sorted(
    precedents,
    key=lambda x: (x["decision_date"], x["precedent_id"])
)

position = {
    p["precedent_id"]: idx
    for idx, p in enumerate(precedents_sorted)
}

citation_edges = []

for idx, precedent in enumerate(precedents_sorted):
    eligible = precedents_sorted[:idx]

    if not eligible:
        continue

    cited_count = random.randint(1, min(5, len(eligible)))
    cited = random.sample(eligible, cited_count)

    precedent["authorities_cited"] = [
        item["precedent_id"]
        for item in cited
    ]

    distinguished = []
    criticized = []

    if cited and random.random() < 0.35:
        distinguished = [
            random.choice(cited)["precedent_id"]
        ]

    if cited and random.random() < 0.15:
        criticized = [
            random.choice(cited)["precedent_id"]
        ]

    precedent["authorities_distinguished"] = distinguished
    precedent["authorities_criticized"] = criticized

    for target in cited:
        relation = "cites"

        if target["precedent_id"] in distinguished:
            relation = "distinguishes"

        if target["precedent_id"] in criticized:
            relation = "criticizes"

        if precedent["treatment_status"] == "Overruled" and random.random() < 0.20:
            relation = "overrules"

        citation_edges.append({
            "source_precedent_id": precedent["precedent_id"],
            "target_precedent_id": target["precedent_id"],
            "relationship": relation
        })

assert len(citation_edges) > 1000
print("Citation and treatment edges:", len(citation_edges))
```

## Ontology indexes and maps

The vault creates navigable pages for:

- courts;
- jurisdictions;
- doctrines;
- legal issues;
- active matters;
- precedent treatments;
- citation relationships.

The purpose is to expose relationships that would otherwise remain buried in prose.


```python
precedents_by_court = defaultdict(list)
precedents_by_jurisdiction = defaultdict(list)
precedents_by_doctrine = defaultdict(list)
precedents_by_issue = defaultdict(list)

for precedent in precedents:
    precedents_by_court[precedent["court"]].append(precedent)
    precedents_by_jurisdiction[precedent["jurisdiction"]].append(precedent)
    precedents_by_doctrine[precedent["cause_of_action"]].append(precedent)
    precedents_by_issue[precedent["legal_issue"]].append(precedent)

for court, items in precedents_by_court.items():
    safe = court.replace("/", "-")
    lines = [f"# {court}", "", f"Precedents: **{len(items)}**", ""]
    lines += [
        f"- [[../01_Precedents/{p['precedent_id']}]] — {p['caption']}"
        for p in sorted(items, key=lambda x: x["decision_date"], reverse=True)
    ]
    write_note(VAULT / "03_Courts" / f"{safe}.md", lines)

for jurisdiction, items in precedents_by_jurisdiction.items():
    lines = [f"# {jurisdiction}", "", f"Precedents: **{len(items)}**", ""]
    lines += [
        f"- [[../01_Precedents/{p['precedent_id']}]] — {p['legal_issue']}"
        for p in sorted(items, key=lambda x: x["decision_date"], reverse=True)
    ]
    write_note(VAULT / "04_Jurisdictions" / f"{jurisdiction}.md", lines)

for doctrine, items in precedents_by_doctrine.items():
    safe = doctrine.replace("/", "-")
    lines = [f"# {doctrine}", "", f"Precedents: **{len(items)}**", ""]
    lines += [
        f"- [[../01_Precedents/{p['precedent_id']}]] — "
        f"{p['jurisdiction']} — {p['legal_issue']}"
        for p in sorted(items, key=lambda x: x["decision_date"], reverse=True)
    ]
    write_note(VAULT / "05_Doctrines" / f"{safe}.md", lines)

for issue, items in precedents_by_issue.items():
    safe = issue.replace("/", "-")
    lines = [f"# {issue}", "", f"Precedents: **{len(items)}**", ""]
    lines += [
        f"- [[../01_Precedents/{p['precedent_id']}]] — "
        f"{p['jurisdiction']} — {p['precedential_status']}"
        for p in sorted(items, key=lambda x: x["decision_date"], reverse=True)
    ]
    write_note(VAULT / "06_Issues" / f"{safe}.md", lines)

print("Court notes:", len(list((VAULT / '03_Courts').glob('*.md'))))
print("Jurisdiction notes:", len(list((VAULT / '04_Jurisdictions').glob('*.md'))))
print("Doctrine notes:", len(list((VAULT / '05_Doctrines').glob('*.md'))))
print("Issue notes:", len(list((VAULT / '06_Issues').glob('*.md'))))
```

## Validation

Baby Step 0 is complete only if:

- exactly 1,000 precedent notes exist;
- exactly five active-matter notes exist;
- all identifiers are unique;
- all precedents are labeled synthetic;
- all active matters remain recommendation-free;
- no recommendation files exist;
- citation relationships point to valid precedent identifiers;
- dashboard, ontology, data dictionary, hot cache, decision, and datasets exist.


```python
errors = []

precedent_notes = list((VAULT / "01_Precedents").glob("*.md"))
matter_notes = list((VAULT / "02_Active_Matters").glob("*.md"))
recommendation_files = list((VAULT / "08_Recommendations").glob("*"))

if len(precedent_notes) != 1000:
    errors.append(
        f"Expected 1000 precedent notes, found {len(precedent_notes)}"
    )

if len(matter_notes) != 5:
    errors.append(
        f"Expected 5 active-matter notes, found {len(matter_notes)}"
    )

precedent_ids = {
    p["precedent_id"]
    for p in precedents
}

matter_ids = {
    m["matter_id"]
    for m in ACTIVE_MATTERS
}

if len(precedent_ids) != 1000:
    errors.append("Precedent identifiers are not unique")

if len(matter_ids) != 5:
    errors.append("Matter identifiers are not unique")

if recommendation_files:
    errors.append(
        "Recommendation folder is not empty at Baby Step 0"
    )

if any(not p["synthetic"] for p in precedents):
    errors.append("One or more precedents lack synthetic labeling")

if any(
    m["recommendation_status"] != "Not Yet Developed"
    for m in ACTIVE_MATTERS
):
    errors.append("One or more matters contain a recommendation")

for edge in citation_edges:
    if edge["source_precedent_id"] not in precedent_ids:
        errors.append(
            f"Invalid citation source: {edge['source_precedent_id']}"
        )
    if edge["target_precedent_id"] not in precedent_ids:
        errors.append(
            f"Invalid citation target: {edge['target_precedent_id']}"
        )

required_files = [
    VAULT / "00_System" / "Legal_Ontology.json",
    VAULT / "00_System" / "Data_Dictionary.md",
    VAULT / "09_Decisions" / "DEC-000.md",
    VAULT / "10_Reports" / "Dashboard.md",
    VAULT / "10_Reports" / "Precedent_Index.md",
    VAULT / "10_Reports" / "Active_Matter_Index.md",
    VAULT / "12_Hot_Cache" / "Current_State.md",
    DATA / "precedents.csv",
    DATA / "precedents.json",
    DATA / "active_matters.csv",
    DATA / "active_matters.json",
    DATA / "citation_edges.csv",
    DATA / "citation_edges.json"
]

for path in required_files:
    if not path.exists():
        errors.append(f"Missing required file: {path}")

combined_bytes = (
    (DATA / "precedents.json").read_bytes()
    + (DATA / "active_matters.json").read_bytes()
    + (DATA / "citation_edges.json").read_bytes()
)

dataset_hash = hashlib.sha256(combined_bytes).hexdigest()

validation = {
    "validated_at": datetime.datetime.now().isoformat(),
    "canonical_vault": str(VAULT),
    "precedent_count": len(precedents),
    "active_matter_count": len(ACTIVE_MATTERS),
    "citation_edge_count": len(citation_edges),
    "recommendation_file_count": len(recommendation_files),
    "combined_sha256": dataset_hash,
    "errors": errors,
    "passed": len(errors) == 0
}

(VAULT / "11_Audit" / "Baby_Step_0_Validation.json").write_text(
    json.dumps(validation, indent=2),
    encoding="utf-8"
)

validation_note = [
    "# Vault Validation Report",
    "",
    f"**Passed:** {validation['passed']}",
    f"**Precedents:** {validation['precedent_count']}",
    f"**Active matters:** {validation['active_matter_count']}",
    f"**Citation/treatment edges:** {validation['citation_edge_count']}",
    f"**Recommendation files:** {validation['recommendation_file_count']}",
    f"**Combined SHA-256:** `{dataset_hash}`",
    "",
    "## Errors",
    ""
]

if errors:
    validation_note += [
        f"- {error}"
        for error in errors
    ]
else:
    validation_note.append("- None")

write_note(
    VAULT / "10_Reports" / "Vault_Validation_Report.md",
    validation_note
)

assert validation["passed"], errors[:20]

print(json.dumps(validation, indent=2))
print("BABY STEP 0 PASSED")
```
